```python
import os
import json
import time
from typing import List, Tuple, Dict, Any, Optional, Union
import pytz
from datetime import datetime

import pandas as pd
from tqdm import tqdm
import gspread
from google.oauth2.service_account import Credentials
from gspread_dataframe import get_as_dataframe, set_with_dataframe

import cloudvolume as cv
import caveclient
import re
import pymaid

path = 'D:/flywire_backup/cave/cat_client.json' # replace with your path

with open(path, 'r') as f:
    conn_file = json.load(f)
api_token = conn_file['api_token']
http_user = conn_file['http_user']
http_password = conn_file['http_password']
server = conn_file['server']

cat_client = pymaid.CatmaidInstance(server = server,  # update with your auth
                                api_token = api_token,
                                caching = True,
                                project_id = 8,
                                http_user = http_user,
                                http_password = http_password 
                                )
```

```python
# ---------- USER CONFIG ----------
SPREADSHEET_NAME = 'Megalopta_PB2'

if SPREADSHEET_NAME == 'Megalopta_FBEB':
    SHEET_NAMES = ['FBEB_EPG', 'FBEB_PEN']  # leave blank [] to process all tabs
elif SPREADSHEET_NAME == 'Megalopta_PB2':
    SHEET_NAMES = ['megalopta_pb2_PFN', 'megalopta_pb2_EPG', 'megalopta_pb2_PEN'] #'megalopta_pb2_delta7', 
else:
    SHEET_NAMES = []  # fallback: all tabs

if SPREADSHEET_NAME == 'Megalopta_FBEB':
    SPREADSHEET_ID = '1HblRPtoJ5m__1QSB9DDVX1nlaRi8e0ix7TanzHKgqhw'
elif SPREADSHEET_NAME == 'Megalopta_PB':
    SPREADSHEET_ID = '1aNp3-ukGhYFOaSPB8ODhSjXTeEFbeOwpTO-TZ9Mxn5c'
elif SPREADSHEET_NAME == 'Megalopta_PB2':
    SPREADSHEET_ID = '1BayKkVWsFcjyZqZvfzfni3VvPcigZco47HtrzduoWho'  # UPDATE from copy
else:
    raise ValueError(f"Unknown SPREADSHEET_NAME: {SPREADSHEET_NAME}")

# 2) Column names (must match your Google Sheet headers exactly)
COL_NAME = "Catmaid name"
COL_SKID = "CATMAID skid"

# 3) pymaid config assumed available in your environment:
# from pymaid import get_skids_by_annotation
# cat_client = <your_initialized_remote_instance>

old_name_prefix = 'old_name:'  # matches your annotation scheme

# ---------- AUTH ----------
# Option A: Service account (recommended for automation)
# gc = gspread.service_account(filename='path/to/service_account.json')

# Option B: OAuth (browser-based, stores token locally)
gc = gspread.oauth()  # uncomment if you prefer interactive auth

# ---------- OPEN SHEET ----------
sh = gc.open_by_key(SPREADSHEET_ID)

# If SHEET_NAMES == [], process all tabs
if not SHEET_NAMES:
    SHEET_NAMES = [ws.title for ws in sh.worksheets()]

print(f"Processing spreadsheet '{SPREADSHEET_NAME}' with tabs: {SHEET_NAMES}")

# --- helpers ---
def _norm(s: str) -> str:
    """Normalize a header string: collapse whitespace (incl. NBSP), lowercase."""
    s = "" if s is None else str(s)
    s = s.replace("\u00A0", " ")          # NBSP -> space
    s = re.sub(r"\s+", " ", s.strip())    # collapse spaces/tabs
    return s.casefold()

def _find_header_row(values, expected_header_norm, scan_rows=10):
    """
    Look through the first 'scan_rows' rows to find the one that contains
    the expected header (normalized). Return 0-based row index or None.
    """
    for i, row in enumerate(values[:scan_rows]):
        row_norm = [_norm(x) for x in row]
        if expected_header_norm in row_norm:
            return i
    return None

def load_sheet_as_df(ws, expected_header="Catmaid name"):
    """
    Read a worksheet robustly:
      - fetch raw values
      - auto-detect the header row that contains expected_header (normalized)
      - normalize column names
      - drop fully empty rows/cols
    """
    values = ws.get_all_values()  # raw grid, includes blanks
    if not values:
        return pd.DataFrame()

    hdr_norm = _norm(expected_header)
    header_row_idx = _find_header_row(values, hdr_norm, scan_rows=10)

    # Fallback: assume first row is header if we couldn't detect it
    if header_row_idx is None:
        header_row_idx = 0

    header = values[header_row_idx]
    data = values[header_row_idx + 1 :]

    # Pad rows to header length
    width = len(header)
    data = [row + [""] * (width - len(row)) for row in data]

    df = pd.DataFrame(data, columns=header)

    # Drop fully empty rows/cols
    df.replace("", pd.NA, inplace=True)
    df = df.dropna(how="all").dropna(axis=1, how="all")
    df = df.fillna("")

    # --- normalize column names but preserve originals for writing back ---
    # Keep a mapping from normalized -> original
    norm_map = {_norm(c): c for c in df.columns}
    df.columns = [norm_map.get(_norm(c), c) for c in df.columns]  # idempotent

    return df

def ensure_columns(df, COL_NAME="Catmaid name", COL_SKID="CATMAID skid"):
    """
    Ensure required columns exist; be permissive about header variants by matching normalized names.
    """
    existing_norm = {_norm(c): c for c in df.columns}

    # Resolve COL_NAME
    if _norm(COL_NAME) not in existing_norm:
        # Try to guess a close match (e.g., "catmaid  name", "catmaidname")
        candidates = [c for c in df.columns if "catmaid" in _norm(c) and "name" in _norm(c)]
        if candidates:
            name_col = candidates[0]
        else:
            raise KeyError(f"Required column '{COL_NAME}' not found in sheet. Found columns: {list(df.columns)}")
    else:
        name_col = existing_norm[_norm(COL_NAME)]

    # Resolve/ensure COL_SKID
    if _norm(COL_SKID) in existing_norm:
        skid_col = existing_norm[_norm(COL_SKID)]
    else:
        skid_col = COL_SKID  # create with canonical label
        df[skid_col] = 0

    # Coerce types
    df[skid_col] = pd.to_numeric(df[skid_col], errors="coerce").fillna(0).astype(int)
    df[name_col] = df[name_col].astype(str).fillna("")

    # Return df and the resolved column names (so downstream code uses the actual headers)
    return df, name_col, skid_col

def write_df_back(ws, df):
    """
    Write DataFrame back to the worksheet, replacing the sheet size to match df.
    Preserves column headers.
    """
    # Optionally, sort columns to keep original order if desired
    set_with_dataframe(
        ws,
        df,
        row=1, col=1,
        include_index=False,
        include_column_header=True,
        resize=True
    )

def update_skids(df, name_col, skid_col, old_name_prefix='old_name:'):
    missing_neurons = []
    df[skid_col] = 0
    for n, row in df.iterrows():
        name = str(row[name_col]).strip()
        if not name or name.lower() in {"nan", "none"}:
            df.at[n, skid_col] = 0
            continue
        annotation = f"{old_name_prefix}{name}"
        try:
            skid_list = pymaid.get_skids_by_annotation(annotations=annotation,
                                                       remote_instance=cat_client)
            if not skid_list:
                raise ValueError("No skids returned")
            skid = int(skid_list[0])
            df.at[n, skid_col] = skid
            print(f"[{annotation}] -> skid {skid}")
        except Exception:
            df.at[n, skid_col] = 0
            print(f"[{annotation}] -> no skid found")
            missing_neurons.append(name)
    return df, missing_neurons

# --- MAIN LOOP ---
all_missing = {}
for title in SHEET_NAMES:
    try:
        ws = sh.worksheet(title)
    except gspread.WorksheetNotFound:
        print(f"WARNING: worksheet {title} not found")
        continue

    print(f"\n--- Processing tab: {title} ---")
    df = load_sheet_as_df(ws, expected_header=COL_NAME)
    df, name_col, skid_col = ensure_columns(df, COL_NAME=COL_NAME, COL_SKID=COL_SKID)
    df, missing = update_skids(df, name_col, skid_col)
    write_df_back(ws, df)
    if missing:
        all_missing[title] = missing

# --- REPORT ---
print("\nSummary:")
if all_missing:
    for tab, names in all_missing.items():
        print(f"- {tab}: {len(names)} neurons with no skid found")
        for nm in names:
            print(f"    • {nm}")
else:
    print("All skids found for all processed rows.")
```
